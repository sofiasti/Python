Instalamos las librerías a utilizar

```python
import pandas as pd
from ydata_profiling import ProfileReport
```

Buscamos el dataset

```python
file_path = '/content/churn (1).csv'
data = pd.read_csv(file_path)
```

Generamos el archivo html myeda

```python
profile = ProfileReport(data, title="EDA Automático - Informe de Calidad de Datos", explorative=True)
output_html_path = "myeda.html"
profile.to_file(output_html_path)
```

La primera valoración serán los nulos, donde podremos ver con cuantos de estos campos estamos trabajando

```python
missing_values = data.isnull().sum()
missing_percentage = (missing_values / len(data)) * 100
print("Valores nulos por columna:\n", missing_values)
print("\nPorcentaje de valores nulos:\n", missing_percentage)
```

También la cantidad de filas duplicadas

```python
duplicated_rows = data.duplicated().sum()
print(f"Cantidad de filas duplicadas: {duplicated_rows}")
```

Mas información sobre los datos que estamos manejando como la varianza, maximos y promedios
